```python
sales = float(input("enter sales ($): "))

cogs = 0.4 * sales
gross_profit = sales - cogs

print(f"Sales: {sales}$")
print(f"Cost of Goods sold: {cogs}$")
print(f"Gross profit: {gross_profit}$")

tax_rate = .065
taxes = (gross_profit * tax_rate)
net_income = gross_profit - taxes

print(f"Sales tax: {taxes}$")
print(f"Net income: {net_income}$")
```

```python
num1 = float(input("enter first number: "))
num2 = float(input("enter second number: "))
op = input("Choose Numerical Operator (+, -, *, /, //, **, %): ")

if op == '+':
  print(f"{num1} + {num2} = {num1 + num2}")
elif op == '-':
  print(f"{num1} - {num2} = {num1 - num2}")
elif op == '*':
  print(f"{num1} * {num2} = {num1 * num2}")
elif op == '/':
  print(f"{num1} / {num2} = {num1 / num2}")
elif op == '//':
  print(f"{num1} // {num2} = {num1 // num2}")
elif op == '**':
  print(f"{num1} ** {num2} = {num1 ** num2}")
elif op == '%':
  print(f"{num1} % {num2} = {num1 % num2}")
else:
  print("Invalid Operation")
```

```python
base_fee = 5.0
hourly_rate = 2.50

print("Hours\tFee")
for hours in range(0,9):
  total_fee = base_fee + (hourly_rate * hours)
  if total_fee < 10.0:
    total_fee = 10.0
  elif total_fee > 20.0:
    total_fee = 20.0
  print(f"{hours}\t${total_fee}")
```

```python
base_fee = 5.0
hourly_rate = 2.50

print("Hours\tFee")
hours = 0
while hours <= 8:
  total_fee = base_fee + (hourly_rate * hours)
  if total_fee < 10.0:
    total_fee = 10.0
  elif total_fee > 20.0:
    total_fee = 20.0
  print(f"{hours}\t${total_fee}")
  hours += 1
```

```python
import math

ticket_price = 10
initAttendees = 20

print("Ad$\tProfit")

for ad in range(0, 201, 25):
  revenue = (initAttendees + 2 * round(math.sqrt(ad))) * ticket_price
  costs = 200 + ad
  profit = revenue - costs
  print(f"{float(ad)}\t{float(profit)}")
```

```python
import math

ticket_price = 10
initAttendees = 20

print("Ad$\tProfit")

ad = 0
while ad <= 200:
  revenue = (initAttendees + 2 * round(math.sqrt(ad))) * ticket_price
  costs = 200 + ad
  profit = revenue - costs
  print(f"{float(ad)}\t{float(profit)}")
  ad += 25
```
